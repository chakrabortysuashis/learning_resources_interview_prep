Langchain consists of many components that can be used to build applications. 
The famous lanhchain components include LLMs, Prompts, Agents, Chains and Memory.
Runnables are another important component of Langchain that allow developers to define 
and execute units of work in a structured way.
One of the most important components is the Runnable.
Runnables are a way to define a unit of work that can be executed in a specific context.
They can be used to encapsulate logic, manage state, and handle inputs and outputs.


The main hing which Runnables solve is the different componenets of langchian have different way of working and
different ways of execution or different invocation.Like LLM have predict method, Agents have run method, 
Chains have run method and Prompts have format method.
Runnables provide a unified interface for executing these components, making it easier to build complex applications.

```python
from abc import ABC, abstractmethod
```

```python
class Runnable(ABC):

  @abstractmethod
  def invoke(input_data):
    pass
```

This is a simple implementation of a Runnable that simulates a language model (LLM).Previously LLMs had a predict method, but now they have an invoke method. This is because the invoke method is more general and can be used for any of input data, not just text. The invoke method takes in a prompt and returns a response.The response is generated by randomly selecting one of three predefined responses.

```python
import random

class NakliLLM(Runnable):

  def __init__(self):
    print('LLM created')

  def invoke(self, prompt):
    response_list = [
        'Delhi is the capital of India',
        'IPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}


  def predict(self, prompt):

    response_list = [
        'Delhi is the capital of India',
        'IPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}
```

This is a simple implementation of a Runnable that simulates a prompt template.
Previously prompts had a format method, but now they have an invoke method.
This is because the invoke method is more general and can be used for any.
This helps us to chain different components together and have a unified interface for executing them.
The invoke method takes in a dictionary of input variables and returns a formatted string.

```python
class NakliPromptTemplate(Runnable):

  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def invoke(self, input_dict):
    return self.template.format(**input_dict)

  def format(self, input_dict):
    return self.template.format(**input_dict)
```

This is a simple implementation of a Runnable that simulates an output parser.
This is because the invoke method is more general and can be used for any.
what it does is that from the response of the LLM it extracts the response and returns it.

```python
class NakliStrOutputParser(Runnable):

  def __init__(self):
    pass

  def invoke(self, input_data):
    return input_data['response']
```

This is Runnable connector class it has the invoke method which acts differently this invoke method takes in mutiplt runnable
list objects and then run one after the other where the output of the current runnable object is passed to the other as input and so on.This continues and creates a pipeline. Here the runnable list can consist of anything like the model , prompt , or the str output paresres or the agent etc .

```python
class RunnableConnector(Runnable):

  def __init__(self, runnable_list):
    self.runnable_list = runnable_list

  def invoke(self, input_data):

    for runnable in self.runnable_list:
      input_data = runnable.invoke(input_data)

    return input_data
```

```python
template = NakliPromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length', 'topic']
)
```

```python
llm = NakliLLM()
```

```python
parser = NakliStrOutputParser()
```

```python
chain = RunnableConnector([template, llm, parser])
```

```python
chain.invoke({'length':'long', 'topic':'india'})
```

```python
template1 = NakliPromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)
```

```python
template2 = NakliPromptTemplate(
    template='Explain the following joke {response}',
    input_variables=['response']
)
```

```python
llm = NakliLLM()
```

```python
parser = NakliStrOutputParser()
```

```python
chain1 = RunnableConnector([template1, llm])
```

```python
chain2 = RunnableConnector([template2, llm, parser])
```

```python
final_chain = RunnableConnector([chain1, chain2])
```

```python
final_chain.invoke({'topic':'cricket'})
```
